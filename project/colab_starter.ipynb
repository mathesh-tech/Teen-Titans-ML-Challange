**Step 3: Check the Datasets**

```python
import pandas as pd

# Load all 4 datasets
df1 = pd.read_csv('train_source1.tsv', sep='\t', dtype=str)
df2 = pd.read_csv('train_source2.tsv', sep='\t', dtype=str)
df3 = pd.read_csv('train_source3.tsv', sep='\t', dtype=str)
gt  = pd.read_csv('train_ground_truth.tsv', sep='\t', dtype=str)

print('=== Source 1 ===')
print(f'Shape: {df1.shape}')
print(f'Columns: {df1.columns.tolist()}')
print(df1.head(3))

print('\n=== Source 2 ===')
print(f'Shape: {df2.shape}')
print(f'Columns: {df2.columns.tolist()}')
print(df2.head(3))

print('\n=== Source 3 ===')
print(f'Shape: {df3.shape}')
print(f'Columns: {df3.columns.tolist()}')
print(df3.head(3))

print('\n=== Ground Truth ===')
print(f'Shape: {gt.shape}')
print(f'Columns: {gt.columns.tolist()}')
print(gt.head(3))
```

**Step 4: Missing Values & Country Analysis**

```python
for name, df in [('Source1', df1), ('Source2', df2), ('Source3', df3)]:
    print(f'--- {name} ---')
    print(f'Total records: {len(df)}')
    if 'country' in df.columns:
        print(f'Unique countries: {df.country.nunique()} -> {df.country.unique().tolist()}')
    if 'name' in df.columns:
        miss = df.name.isnull().sum()
        print(f'Missing names: {miss} ({miss/len(df)*100:.1f}%)')
    if 'address' in df.columns:
        miss = df.address.isnull().sum()
        print(f'Missing addresses: {miss} ({miss/len(df)*100:.1f}%)')
    print()
```
